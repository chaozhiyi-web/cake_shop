```python
# @title

# (7,4) Hamming Code

# Step 1: Enter the Data
data = input("Enter 4-bit data: ")

if len(data) != 4 or any(bit not in "01" for bit in data):
    print("Please enter exactly 4 binary bits.")
    exit()

# Step 2: Calculate redundant bits
r = 3
print("Number of redundant bits:", r)

# Step 3: Determine parity bits
code = [0] * 8  # Use positions 1 to 7

code[3] = int(data[0])
code[5] = int(data[1])
code[6] = int(data[2])
code[7] = int(data[3])

# Even parity calculation
code[1] = code[3] ^ code[5] ^ code[7]
code[2] = code[3] ^ code[6] ^ code[7]
code[4] = code[5] ^ code[6] ^ code[7]

print("Original data:", data)
print("Hamming code:", ''.join(map(str, code[1:])))

# Step 4: Create error data for testing
error_pos = int(input("Enter error position (0-7, 0 = no error): "))

if error_pos < 0 or error_pos > 7:
    print("Invalid position!")
    exit()

if error_pos != 0:
    code[error_pos] ^= 1

received = code.copy()

print("Received data:", ''.join(map(str, received[1:])))

# Step 5: Check for errors
s1 = received[1] ^ received[3] ^ received[5] ^ received[7]
s2 = received[2] ^ received[3] ^ received[6] ^ received[7]
s4 = received[4] ^ received[5] ^ received[6] ^ received[7]

error = s1 + (s2 * 2) + (s4 * 4)

if error == 0:
    print("No error detected!")
else:
    print("Error detected at position:", error)

    # Step 6: Correct the error
    received[error] ^= 1

    print("Corrected data:", ''.join(map(str, received[1:])))
```
